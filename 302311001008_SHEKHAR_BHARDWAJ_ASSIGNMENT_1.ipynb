## Loading the datasets

### Subtask:
Load the Iris and Breast Cancer datasets from the UCI repository.


**Reasoning**:
Import the necessary function to load datasets from scikit-learn and load both the Iris and Breast Cancer datasets.



```python
from sklearn.datasets import load_iris, load_breast_cancer

iris = load_iris()
breast_cancer = load_breast_cancer()

print("Iris dataset loaded successfully.")
print("Breast Cancer dataset loaded successfully.")
```

## Prepare the data

### Subtask:
Splitting the datasets into training and testing sets.


**Reasoning**:
Importing the necessary function and splitting both datasets into training and testing sets as instructed.



```python
from sklearn.model_selection import train_test_split

test_sizes = [0.3, 0.4, 0.5, 0.2, 0.1] # Corresponding to 30/70, 40/60, 50/50, 20/80, 10/90 splits

data_splits = {}

for test_size in test_sizes:
    # Iris dataset
    X_train_iris, X_test_iris, y_train_iris, y_test_iris = train_test_split(
        iris.data, iris.target, test_size=test_size, random_state=42
    )
    data_splits[f'iris_{test_size}'] = {
        'X_train': X_train_iris, 'X_test': X_test_iris,
        'y_train': y_train_iris, 'y_test': y_test_iris
    }

    # Breast Cancer dataset
    X_train_bc, X_test_bc, y_train_bc, y_test_bc = train_test_split(
        breast_cancer.data, breast_cancer.target, test_size=test_size, random_state=42
    )
    data_splits[f'bc_{test_size}'] = {
        'X_train': X_train_bc, 'X_test': X_test_bc,
        'y_train': y_train_bc, 'y_test': y_test_bc
    }

    print(f"Datasets split successfully with test size: {test_size}")
```

## Training and evaluating decision tree classifiers



**Reasoning**:
Importing the necessary libraries for Decision Tree classification, parameter tuning, evaluation metrics, and plotting. Defining the parameter grids for GridSearchCV for both datasets. We Calculate and print the evaluation metrics for both datasets. Generating and we display the decision tree plots for the best models.



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.datasets import load_iris, load_breast_cancer

# Load UCI datasets via sklearn
iris = load_iris()
breast_cancer = load_breast_cancer()

# Test size splits to evaluate
test_sizes = [0.3, 0.4, 0.5, 0.2, 0.1]  # 30/70, 40/60, etc.

# Store results for analysis
results = {}

for test_size in test_sizes:
    print(f"\n========== Test Size: {test_size} ==========\n")

    # Loop through both datasets
    for name, dataset in [('Iris', iris), ('Breast Cancer', breast_cancer)]:
        X = dataset.data
        y = dataset.target

        # Split data
        X_train, X_test, y_train, y_test = train_test_split(
            X, y, test_size=test_size, random_state=42
        )

        # Train Decision Tree
        classifier = DecisionTreeClassifier(random_state=42)
        classifier.fit(X_train, y_train)
        y_pred = classifier.predict(X_test)

        # Evaluation
        print(f"--- {name} Dataset ---")
        print("Confusion Matrix:")
        print(confusion_matrix(y_test, y_pred))
        print("\nClassification Report:")
        print(classification_report(y_test, y_pred))

        # Store performance in dictionary
        results[f'{name}_{test_size}'] = {
            'classifier': classifier,
            'accuracy': classifier.score(X_test, y_test),
            'X_test': X_test,
            'y_test': y_test,
            'y_pred': y_pred
        }

        # Plot tree for only the smallest test size for readability
        if test_size == 0.2:
            plt.figure(figsize=(15, 10))
            plot_tree(
                classifier,
                feature_names=dataset.feature_names,
                class_names=dataset.target_names,
                filled=True
            )
            plt.title(f"Decision Tree - {name} Dataset (test size={test_size})")
            plt.show()

# Summary of accuracies
print("\n===== Accuracy Summary =====")
for key, val in results.items():
    print(f"{key}: {val['accuracy']:.4f}")
```
